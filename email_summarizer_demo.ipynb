```python
import os
from google import genai
from google.genai import types
from getpass import getpass

os.environ['GEMINI_API_KEY'] = getpass('Gemini API key: ')
client = genai.Client(api_key=os.environ['GEMINI_API_KEY'])
```

```python
emails = [
    {'id':'1','subject':'Interview invitation','body':'Please confirm a time slot next week.'},
    {'id':'2','subject':'Invoice #2043','body':'Payment is due on Friday.'},
    {'id':'3','subject':'Order question','body':'Can you tell me when my order will arrive?'}
]

prompt = '''Summarize each email, say whether action is required, suggest the action, and assign urgency. Return JSON.\n\n''' + str(emails)

response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
    config=types.GenerateContentConfig(response_mime_type='application/json', temperature=0.2)
)

print(response.text)
```
